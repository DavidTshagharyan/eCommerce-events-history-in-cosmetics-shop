```python
# Bring user_features և session_features tables DataFrame
user_features = pd.read_sql("SELECT * FROM user_features;", engine)
session_features = pd.read_sql("SELECT * FROM session_features;", engine)

print("User features shape:", user_features.shape)
print("Session features shape:", session_features.shape)

user_features.head()
```

```python
# Overall conversion rate
buyer_rate = user_features['is_buyer'].mean()
session_conversion_rate = session_features['converted'].mean()

print(f"User-level buyer rate: {buyer_rate:.2%}")
print(f"Session-level conversion rate: {session_conversion_rate:.2%}")
```

```python
# Ստուգենք extreme session duration-ները
extreme_sessions = session_features[session_features['session_duration_min'] > 1440]  # ավելի քան 1 օր
print(f"Sessions longer than 1 day: {len(extreme_sessions)}")
print(f"Percentage: {len(extreme_sessions) / len(session_features):.4%}")

# Ստուգենք user-level outlier-ները (bot-like users)
suspicious_users = user_features[user_features['frequency_sessions'] > 100]
print(f"\nUsers with >100 sessions: {len(suspicious_users)}")
print(f"Percentage of total users: {len(suspicious_users) / len(user_features):.4%}")

# Ինչպիսի՞ն է normal session-ի տևողությունը (առանց outlier-ների)
normal_sessions = session_features[session_features['session_duration_min'] <= 1440]
print(f"\nNormal sessions - median duration (min): {normal_sessions['session_duration_min'].median():.2f}")
print(f"Normal sessions - 95th percentile duration (min): {normal_sessions['session_duration_min'].quantile(0.95):.2f}")
```

```python
# 1. Մաքուր session_features (հեռացնում ենք >24h sessions)
session_features_clean = session_features[session_features['session_duration_min'] <= 1440].copy()

print(f"Removed {len(session_features) - len(session_features_clean)} extreme sessions")
print(f"New shape: {session_features_clean.shape}")

# 2. User-features-ում ավելացնենք "is_suspicious" flag (bot-like users)
user_features['is_suspicious'] = (user_features['frequency_sessions'] > 100).astype(int)

print(f"\nFlagged {user_features['is_suspicious'].sum()} suspicious users")

# 3. Ստուգենք նոր conversion rate-ը մաքուր տվյալի վրա
new_conversion_rate = session_features_clean['converted'].mean()
print(f"\nSession conversion rate (cleaned): {new_conversion_rate:.2%}")
```

Saving for SQL Queries

```python
session_features_clean.to_sql('session_features_clean', engine, if_exists='replace', index=False)
user_features.to_sql('user_features_clean', engine, if_exists='replace', index=False)

print("Saved to PostgreSQL ✅")
```

Saving CSV Local

```python
session_features_clean.to_csv('../data/processed/session_features_clean.csv', index=False)
user_features.to_csv('../data/processed/user_features_clean.csv', index=False)

print("Saved to CSV ✅")
```
